```python
df=df.fillna(df.median(numeric_only=True))
df=df.dropna()
print((df.isnull().sum()*100/(df.shape[0])).sort_values())
```

```python
df.shape
df["RainTomorrow"]=df["RainTomorrow"].apply(lambda x: 1 if x == "Yes" else 0)
df["RainToday"]=df["RainToday"].apply(lambda x: 1 if x == "Yes" else 0)

df.info()
```

```python


dirs = ['N','NNE','NE','ENE','E','ESE','SE','SSE','S','SSW','SW','WSW','W','WNW','NW','NNW']
dir_to_deg = {d: i*(360/16) for i, d in enumerate(dirs)}

for col in ['WindGustDir', 'WindDir9am', 'WindDir3pm']:
    df[col] = df[col].map(dir_to_deg)
    df[col + '_sin'] = np.sin(np.deg2rad(df[col]))
    df[col + '_cos'] = np.cos(np.deg2rad(df[col]))
    df.drop(col, axis=1, inplace=True)

```

```python
df = pd.get_dummies(df, columns=['Location'], prefix='Loc',drop_first=True,dtype=int)

df.head()
```

```python
trig_cols = [col for col in df.columns if col.endswith(('_sin', '_cos'))]
exclude_cols=[col for col in df.columns if set(df[col].unique()).issubset({0,1})]
exclude_cols += trig_cols

to_standardize_cols = [c for c in df.columns if c not in exclude_cols]

print(to_standardize_cols)
print(exclude_cols)
```

```python
Rainfall_Today    = df["RainToday"]
Rainfall_Tomorrow = df["RainTomorrow"]

X = df.drop(columns=["RainToday", "RainTomorrow"])

n = len(X)
train_size = int(0.8 * n)

# ---- ONLY CHANGE STARTS HERE ----
np.random.seed(1)
perm = np.random.permutation(n)

train_idx = perm[:train_size]
test_idx  = perm[train_size:]
# ---- ONLY CHANGE ENDS HERE ----
X1 = X.iloc[train_idx].copy()
X2 = X.iloc[test_idx].copy()
Y1 = Rainfall_Today.iloc[train_idx].to_numpy().reshape(-1, 1)
Y2 = Rainfall_Today.iloc[test_idx].to_numpy().reshape(-1, 1)

Z1 = Rainfall_Tomorrow.iloc[train_idx].to_numpy().reshape(-1, 1)
Z2 = Rainfall_Tomorrow.iloc[test_idx].to_numpy().reshape(-1, 1)

mean = X1[to_standardize_cols].mean()
std  = X1[to_standardize_cols].std()

X1[to_standardize_cols] = (X1[to_standardize_cols] - mean) / std
X2[to_standardize_cols] = (X2[to_standardize_cols] - mean) / std


```

```python
def predict(X,W,b):
    return sigmoid(X@W+b)
def sigmoid(z):
    return 1 / (1 + np.exp(-z))
def binary_cross_entropy(X, y, W, b, eps=1e-15):          
    p = predict(X,W,b)
    p = np.clip(p, eps, 1 - eps)

    return -np.mean(
        y * np.log(p) + (1 - y) * np.log(1 - p)
    )

def compute_gradient(X,Y,W,b):
    n=X.shape[0]
    Y_pred=predict(X,W,b)
    E=Y_pred-Y
    dW=(1/n)*X.T@E
    db = (1/n) * np.sum(E)
    return dW,db
def update(W,b,dW,db,lr):
    W=W-lr*dW
    b=b-lr*db
    return W,b
def accuracy(y, p):
    return np.mean((p >= 0.5) == y)

def train(X,Y,X2,Y2,lr=0.01,epochs=10,I_O=0):
    X=np.array(X)
    Y=np.array(Y).reshape(-1,1)
    m=X.shape[1]
    W=np.zeros((m,1))
    b=0.0
    losses = []
    for i in range(epochs):
        dW,db=compute_gradient(X,Y,W,b)
        W,b=update(W,b,dW,db,lr)
        Y_pred = predict(X ,W, b)
        loss = binary_cross_entropy(X, Y, W, b, eps=1e-15)
        if I_O==0:
            continue
        if i%1000==0:
            Y2_pred = predict(X2, W, b)
            acc_train = accuracy(Y1,Y_pred)
            acc_test = accuracy(Y2,Y2_pred)
            losses.append(loss)


            print("Accuracy(%):", acc_train * 100)
            print("Accuracy(%):", acc_test * 100)


    return W,b,losses
```

```python
W,b,losses=train(X1,Y1,X2,Y2,epochs=5000,I_O=1)
```

```python
print("Rain frequency:", 100*np.mean(Y1))
print("Model accuracy:", 100*accuracy(Y1, predict(X1, W, b)))
print(W.ravel(),b,losses)
```
